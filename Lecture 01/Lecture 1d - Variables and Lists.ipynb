# <span style="color:darkblue"> Visualizing Lists </span>

<font size="5"> 

A histogram for categorical data

```python
list_list = list_colors + ['red']
```

```python
# This creates a histogram with the "list_colors"
plt.hist(x= list_list)
```

```python
# Creat a list with repeated string values
# Compute your own histogram

list_created_by_me_2 = ["yes", "yes", "no", "yes", "no", "yes"]
plt.hist(x = list_created_by_me_2)
```

<font size="5"> 

A scatterplot for continuous data

```python
print(list(list_numbers))
print(list_numbers_sqr)
```

```python
type(list_numbers)
```

```python
plt.scatter(x = list_numbers, y = list_numbers_sqr)
plt.xlabel("A meaningful name for the X-axis")
plt.ylabel("Favourite name for Y-axis")
plt.show()
```

```python
# Create a list with numbers, then create your own scatter plot

list_created_by_me_3 = [2,3,4,5,6,7,8,9,100]
list_created_by_me_4 = [77,20,19,18,17,16,15,14,13]

plt.scatter(x = list_created_by_me_3, y = list_created_by_me_4)
plt.xlabel("A series of random numbers")
plt.ylabel("Another series of random numbers")
plt.show()
```
