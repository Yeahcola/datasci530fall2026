````python
# We will "alias" two sublibraries in "statsmodels"
# "statsmodels.formula.api" contains functions to estimate models
# "statsmodels.api" contains general-use statistical options

import statsmodels.api as sm
import statsmodels.formula.api as smf
import maketables as mt
````

# <span style="color:darkblue"> II. Generate Simulated Data </span>

<font size = "5">

Create an empty dataset

````python
dataset = pd.DataFrame()
````

<font size = "5">

Create four random variables of size ($n = 100$)

````python
n = 100
np.random.seed(42)
dataset["x"] = np.random.normal(loc = 0,scale = 1, size = n)
dataset["z"] = np.random.normal(loc = 0,scale = 1, size = n)
dataset["e"] = np.random.normal(loc = 0,scale = 1, size = n)
dataset["z with spaces"] = dataset["z"]
````

<font size = "5">

Create discrete random variable ($n = 100$)

````python
dataset["d"] = np.random.choice(a = [1,2,3],
                                size = n,
                                p = [0.2,0.2,0.6])
````

<font size = "5">

Create data from the linear model

$ y = 2 + 5 x + e$

````python
# We can compute formulas directly over dataset columns
dataset["y"] =2 + 5* dataset["x"] + dataset["x"]*dataset["e"]
````

# <span style="color:darkblue"> III. Regression Tables </span>


<font size = "5">

Summaries for univariate regression

````python
# Run the model with multiple variables by using "+"
results_univariate = smf.ols(formula = 'y ~ x',data = dataset).fit(cov_type= "HC1")

# ETable creates a publication-style table from fitted model results.
# The .make(type="gt") command is important when you want to render inside the Python
# notebook.
table = mt.ETable([results_univariate], model_heads=["Univariate"]).make(type="gt")

table
````

<font size = "5">

Detailed diagnostics for univariate regression

````python
print(results_univariate.summary())
````

<font size = "5">

Summaries for multivariate regression

````python
# Run the model with multiple variables by using "+"
results_multivariate = smf.ols(formula = 'y ~ x + z',
                               data = dataset).fit(cov_type = "HC1")

mt.ETable([results_multivariate], model_heads=["Multivariate"]).make(type="gt")
````

<font size = "5">

Summaries for multivariate regression + colnames w/ spaces

````python
# Use Q("...") to reference variables that
# have spaces in the name
results_multivariate_spaces = smf.ols(formula = 'y ~ x + Q("z with spaces") ',
                                      data = dataset).fit(cov_type = "HC1")
mt.ETable([results_multivariate_spaces], model_heads=["Named variable"]).make(type="gt")
````

<font size = "5">

Summaries for multivariate regression + categories

````python
# Run the model with multiple variables by using "+"
# This creates a set of distinct indicator variables for each category
results_multivariate_category = smf.ols(formula = 'y ~ x + C(d)',
                                        data = dataset).fit(cov_type = "HC1")

# The results are reported with a base category, T.1
mt.ETable([results_multivariate_category], model_heads=["Categorical"]).make(type="gt")
````

<font size = "5">

Summaries for multivariate regression + interaction

````python
# Run the model with multiple variables by using "+"
# This creates a set of distinct indicator variables for each category
results_multivariate_interaction = smf.ols(formula = 'y ~ x + z + z:x',
                                        data = dataset).fit(cov_type = "HC1")

# The results are reported with a base category, T.1
mt.ETable([results_multivariate_interaction], model_heads=["Interaction"]).make(type="gt")
````

# <span style="color:darkblue"> IV. Professional Tables </span>


<font size = "5">

Summaries for multiple columns

````python
list_results = [results_univariate,
                results_multivariate,
                results_multivariate_category,
                results_multivariate_interaction]

mt.ETable(list_results,
          model_heads=["Univariate", "Multivariate", "Categorical", "Interaction"]).make(type="gt")
````

<font size = "5">

Summaries for multiple columns (sorted + titled + stats)

````python
# This list inputs the headings of the table
list_headings   = ["Univariate",
                   "Multivariate",
                   "Categorical",
                   "Interaction"]

# This is the list of regressor names (if you want a particular order)
list_regressors = ["x",
                   "z",
                   "z:x",
                   "C(d)[T.2]",
                   "C(d)[T.3]"]

# keep controls coefficient order; exact_match=True matches exact names.
mt.ETable(
    list_results,
    model_heads=list_headings,
    keep=list_regressors,
    exact_match=True,
    model_stats=["N", "r2"],
    model_stats_labels={"N": "Observations", "r2": "R-squared"},
    caption="Regression Results",
).make(type="gt")
````

# <span style="color:darkblue"> V. Export Regression Tables to Word </span>

<font size = "5">

`ETable` can save a formatted, editable Microsoft Word document (`.docx`). <br>

- The following examples save a one-model table and a side-by-side comparison table in the notebook's current working directory.
- The tables are stored under the subfolder "results"
- Change `file_name` if you want to save them elsewhere.

````python
# Export a table for one fitted model.
single_model_table = mt.ETable(
    [results_univariate],
    model_heads=["Univariate"],
    caption="Univariate Regression Results",
)

single_model_table.save(
    type="docx",
    file_name="results/regression_univariate.docx",
    show=False,
    replace = True
)
````

````python
# Export several models together, with selected coefficients and model statistics.
comparison_table = mt.ETable(
    list_results,
    model_heads=list_headings,
    keep=list_regressors,
    exact_match=True,
    model_stats=["N", "r2"],
    model_stats_labels={"N": "Observations", "r2": "R-squared"},
    caption="Regression Results",
)

comparison_table.save(
    type="docx",
    file_name="results/regression_models.docx",
    show=False,
    replace = True
)
````
